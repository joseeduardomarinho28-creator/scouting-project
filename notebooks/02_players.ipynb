# Players Exploration

Objective: load clubs and players data, and identify which players
belong to clubs from the Brazilian Série A (competition_id = BRA1).

```python
import pandas as pd
```

## Loading clubs data

```python
df_clubs = pd.read_csv('../data/raw/clubs.csv')
```

## Exploring clubs structure

```python
df_clubs.head()
```

```python
df_clubs.info()
```

## Filtering Brazilian clubs

```python
df_clubs_brazil = df_clubs[df_clubs['domestic_competition_id'] == 'BRA1']
df_clubs_brazil
```

```python
df_clubs_brazil[['name', 'last_season']]
```

```python
df_clubs_brazil.sort_values('name')
```

```python
df_clubs_brazil_current = df_clubs_brazil[df_clubs_brazil['last_season'] == 2025]
df_clubs_brazil_current
```

## Loading players data

```python
df_players = pd.read_csv('../data/raw/players.csv')
```

## Exploring players structure

```python
df_players.head()
```

```python
df_players.columns
```

```python
df_players.info()
```

```python
df_players['last_season'].value_counts()
```

## Merging players with Brazilian clubs

```python
df_players_brazil = pd.merge(df_players, df_clubs_brazil_current, left_on='current_club_id', right_on='club_id')
df_players_brazil
```

```python
df_players_brazil['player_id'].duplicated().sum()
```

```python
df_players_brazil.groupby('current_club_name').size()
```

```python
df_players_brazil['last_season_x'].value_counts()
```

```python
df_players_brazil_active = df_players_brazil[df_players_brazil['last_season_x'] == 2025]
df_players_brazil_active
```

## Conclusion

- Loaded clubs.csv and players.csv
- Filtered clubs to Brazilian Série A, current season (20 clubs)
- Merged players with current Brazilian clubs via current_club_id
- Filtered to active players (last_season == 2025): [X] players
- Result stored in df_players_brazil_active, ready for market value analysis
