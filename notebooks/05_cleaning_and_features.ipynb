# Cleaning and Feature Selection

Objective: load the merged player+stats dataset from notebook 04, drop
redundant/irrelevant columns, and prepare a clean feature set for the
Phase 1 cost-benefit scoring and clustering steps.

```python
import pandas as pd
```

## Loading merged data

```python
df_pws = pd.read_csv('../data/processed/players_with_stats_2024.csv')
```

## Dropping redundant and irrelevant columns

```python
columns_to_drop = [
    # duplicate/redundant merge artifacts
    'name_y', 'name_normalized', 'stats_key', 'name_normalized_stats',
    
    # irrelevant metadata
    'url_x', 'url_y', 'image_url', 'filename', 'agent_name', 'player_code',
    
    # club-level info, redundant with current_club_name (one row per player already)
    'club_id', 'club_code', 'domestic_competition_id', 'total_market_value',
    'squad_size', 'average_age', 'foreigners_number', 'foreigners_percentage',
    'national_team_players', 'stadium_name', 'stadium_seats',
    'net_transfer_record', 'coach_name', 'last_season_y',
    
    # matching process helpers, no longer needed
    'matched', 'used', 'last_season_x',
]

df_clean = df_pws.drop(columns=columns_to_drop)
df_clean.shape
```

```python
df_clean.columns.tolist()
```

## Standardizing column names

```python
df_clean = df_clean.rename(columns={
    'name_x': 'player_name',
    'Jogador': 'stats_player_name',  # redundant with player_name where matched, kept for traceability
    'Pos.': 'stats_position',
    'Time': 'stats_club',
    'Min.': 'minutes_played',
    'Gols': 'goals',
    'Assis.': 'assists',
    'PrgP': 'progressive_passes',
    'PrgC': 'progressive_carries',
    'Tent': 'dribbles_attempted',
    'Suc': 'dribbles_successful',
    'SCA': 'shot_creating_actions',
    'GCA': 'goal_creating_actions',
    'Div': 'tackles',
    'Bloqueios': 'blocks',
    'CrtsA': 'yellow_cards_stats',
    'CrtV': 'red_cards_stats',
})
df_clean.columns.tolist()
```

## Exporting cleaned dataset

```python
df_clean.to_csv('../data/processed/players_final_2024.csv', index=False)
```

## Conclusion

- Loaded the merged roster+stats dataset from notebook 04 (1021 players, 66 columns)
- Dropped 25 redundant/irrelevant columns (merge artifacts, club-level
  metadata duplicated across rows, matching-process helper columns),
  resulting in 40 columns
- Standardized column names to English/snake_case for consistency with
  the rest of the project
- Result exported to data/processed/players_final_2024.csv, ready for
  Phase 1 cost-benefit scoring and clustering
