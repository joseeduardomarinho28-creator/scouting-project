## Normalizing names for matching

```python
import unicodedata

def normalize_name(name):
    if pd.isna(name):
        return name
    name = unicodedata.normalize('NFKD', name)  # separate letter + accent
    name = ''.join(c for c in name if not unicodedata.combining(c))  # remove only the accents
    return name.lower().strip()  # make everything lowercase, no spaces at the ends
```

```python
df_players_brazil['name_normalized'] = df_players_brazil['name_x'].apply(normalize_name)
```

```python
df_stats_by_player['name_normalized'] = df_stats_by_player['Jogador'].apply(normalize_name)
```

## Fuzzy matching with club as tie-breaker

```python
from rapidfuzz import fuzz

def find_best_match(nome, clube, stats_df):
    # filters candidates from the same club (comparison 'contains', both ways,
    # because one club name might be an abbreviation of the other)
    candidates = stats_df[stats_df['Time'].apply(
        lambda t: t.lower() in clube.lower() or clube.lower() in t.lower()
    )]
    
    if candidates.empty:
        return None, 0

    # calculates the similarity score between 'name' and each candidate from the same club
    scores = candidates['Jogador'].apply(lambda c: fuzz.WRatio(nome, c))
    best_idx = scores.idxmax()
    best_score = scores.max()
    
    return candidates.loc[best_idx, 'Jogador'], best_score
```

```python
results = no_match.apply(
    lambda row: find_best_match(row['name_x'], row['current_club_name'], df_stats_by_player),
    axis=1
)

no_match['fuzzy_candidate'] = results.apply(lambda r: r[0])
no_match['fuzzy_score'] = results.apply(lambda r: r[1])
```

### Issue found: duplicate candidate assignment

Reviewing the results above revealed that the same candidate could be
assigned as the "best match" for multiple different players (e.g. three
different names all matched to "Tevis Gabriel Alves Santos" with the same
score). This happens because each row is matched independently, with no
check for whether a candidate was already used. Since a real person can
only be one player, this is a genuine one-to-one matching problem —
addressed in the next section.

## One-to-one fuzzy matching with club filter (corrected approach)

```python
from rapidfuzz import fuzz

# exclude candidates already used in the direct/normalized match (the 401)
already_matched_names = df_players_brazil[
    df_players_brazil['name_normalized'].isin(df_stats_by_player['name_normalized'])
]['name_normalized']

df_stats_available = df_stats_by_player[~df_stats_by_player['name_normalized'].isin(already_matched_names)]

# 1. calculate all possible pairs (unmatched player x candidate from the same club)
pairs = []
for idx, row in no_match.iterrows():
    name = row['name_x']
    club = row['current_club_name']
    candidates = df_stats_available[df_stats_available['Time'].apply(
        lambda t: t.lower() in club.lower() or club.lower() in t.lower()
    )]
    for _, cand_row in candidates.iterrows():
        score = fuzz.WRatio(name, cand_row['Jogador'])
        pairs.append({
            'player_idx': idx,
            'name_x': name,
            'candidate': cand_row['Jogador'],
            'score': score
        })

df_pairs = pd.DataFrame(pairs)
```

```python
# 2. sort by score, highest first
df_pairs = df_pairs.sort_values('score', ascending=False)

# 3. go through in order, only accepting a match if neither the player
# nor the candidate has been used yet
used_players = set()
used_candidates = set()
accepted_matches = []

for _, pair in df_pairs.iterrows():
    if pair['score'] < 81:
        break  # already past the threshold, stop looking
    if pair['player_idx'] in used_players or pair['candidate'] in used_candidates:
        continue  # player or candidate already matched, skip this pair
    accepted_matches.append(pair)
    used_players.add(pair['player_idx'])
    used_candidates.add(pair['candidate'])

df_accepted_matches = pd.DataFrame(accepted_matches)
df_accepted_matches
```

```python
# manually reviewed: these matches were rejected as false positives
# (e.g. same club, but different real people)
rejected_player_idx = [440]  # Marcelo Pitaluga (440) incorrectly matched to Marcelo

df_accepted_matches_reviewed = df_accepted_matches[~df_accepted_matches['player_idx'].isin(rejected_player_idx)]
df_accepted_matches_reviewed
```

```python
from rapidfuzz import process

review_rows = []
for idx, row in still_unmatched.iterrows():
    name = row['name_x']
    top_matches = process.extract(name, df_stats_by_player['Jogador'], scorer=fuzz.WRatio, limit=3)
    review_rows.append({
        'name_x': name,
        'current_club_name': row['current_club_name'],
        'top_candidates': top_matches
    })

df_review = pd.DataFrame(review_rows)
df_review
```
