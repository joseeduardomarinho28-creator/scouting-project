## Normalizing names for matching

```python
import unicodedata

def normalize_name(name):
    if pd.isna(name):
        return name
    name = unicodedata.normalize('NFKD', name)  # separate letter + accent
    name = ''.join(c for c in name if not unicodedata.combining(c))  # remove only the accents
    return name.lower().strip()  # make everything lowercase, no spaces at the ends
```

```python
df_players_brazil['name_normalized'] = df_players_brazil['name_x'].apply(normalize_name)
```

## Fuzzy matching with club as tie-breaker

```python
from rapidfuzz import fuzz

def find_best_match(nome, clube, stats_df):
    # filters candidates from the same club (comparison 'contains', both ways,
    # because one club name might be an abbreviation of the other)
    candidates = stats_df[stats_df['Time'].apply(
        lambda t: t.lower() in clube.lower() or clube.lower() in t.lower()
    )]
    
    if candidates.empty:
        return None, 0

    # calculates the similarity score between 'name' and each candidate from the same club
    scores = candidates['Jogador'].apply(lambda c: fuzz.WRatio(nome, c))
    best_idx = scores.idxmax()
    best_score = scores.max()
    
    return candidates.loc[best_idx, 'Jogador'], best_score
```

```python
results = no_match.apply(
    lambda row: find_best_match(row['name_x'], row['current_club_name'], df_stats_by_player),
    axis=1
)

no_match['fuzzy_candidate'] = results.apply(lambda r: r[0])
no_match['fuzzy_score'] = results.apply(lambda r: r[1])
```

### Issue found: duplicate candidate assignment

Reviewing the results above revealed that the same candidate could be
assigned as the "best match" for multiple different players (e.g. three
different names all matched to "Tevis Gabriel Alves Santos" with the same
score). This happens because each row is matched independently, with no
check for whether a candidate was already used. Since a real person can
only be one player, this is a genuine one-to-one matching problem —
addressed in the next section.

## One-to-one fuzzy matching with club filter (corrected approach)

```python
from rapidfuzz import fuzz

# exclude candidates already used in the direct/normalized match (the 401)
already_matched_names = df_players_brazil[
    df_players_brazil['name_normalized'].isin(df_stats_by_player['name_normalized'])
]['name_normalized']

df_stats_available = df_stats_by_player[~df_stats_by_player['name_normalized'].isin(already_matched_names)]

# 1. calculate all possible pairs (unmatched player x candidate from the same club)
pairs = []
for idx, row in no_match.iterrows():
    name = row['name_x']
    club = row['current_club_name']
    candidates = df_stats_available[df_stats_available['Time'].apply(
        lambda t: t.lower() in club.lower() or club.lower() in t.lower()
    )]
    for _, cand_row in candidates.iterrows():
        score = fuzz.WRatio(name, cand_row['Jogador'])
        pairs.append({
            'player_idx': idx,
            'name_x': name,
            'candidate': cand_row['Jogador'],
            'score': score
        })

df_pairs = pd.DataFrame(pairs)
```

```python
# 2. sort by score, highest first
df_pairs = df_pairs.sort_values('score', ascending=False)

# 3. go through in order, only accepting a match if neither the player
# nor the candidate has been used yet
used_players = set()
used_candidates = set()
accepted_matches = []

for _, pair in df_pairs.iterrows():
    if pair['score'] < 81:
        break  # already past the threshold, stop looking
    if pair['player_idx'] in used_players or pair['candidate'] in used_candidates:
        continue  # player or candidate already matched, skip this pair
    accepted_matches.append(pair)
    used_players.add(pair['player_idx'])
    used_candidates.add(pair['candidate'])

df_accepted_matches = pd.DataFrame(accepted_matches)
df_accepted_matches
```

## Reviewing remaining unmatched players with candidate suggestions

```python
# players still unmatched after direct match + accepted fuzzy matches
fuzzy_matched_idx = df_accepted_matches_reviewed['player_idx'].tolist()
still_unmatched = no_match[~no_match.index.isin(fuzzy_matched_idx)]
still_unmatched = no_match[~no_match.index.isin(fuzzy_matched_idx)].drop(columns=['fuzzy_candidate', 'fuzzy_score'])
still_unmatched.shape
still_unmatched.shape
```

```python
still_unmatched.columns.tolist()
```

```python
from rapidfuzz import process

review_rows = []
for idx, row in still_unmatched.iterrows():
    name = row['name_x']
    last_name = name.split()[-1]  # take the last word of the name (usually the surname)

    # pre-filter: only consider candidates whose name contains the surname
    candidates_pool = df_stats_by_player[df_stats_by_player['Jogador'].str.contains(last_name, case=False, na=False)]

    if not candidates_pool.empty:
        top_matches = process.extract(name, candidates_pool['Jogador'], scorer=fuzz.WRatio, limit=3)
        review_rows.append({
            'name_x': name,
            'current_club_name': row['current_club_name'],
            'top_candidates': top_matches
        })

df_review = pd.DataFrame(review_rows)
df_review
```

```python
with open('../data/processed/review_candidates.txt', 'w', encoding='utf-8') as f:
    for _, row in df_review.iterrows():
        f.write(f"{row['name_x']} ({row['current_club_name']})\n")
        for candidate_name, score, _ in row['top_candidates']:
            f.write(f"    → {candidate_name} (score: {score:.1f})\n")
        f.write("\n")

print("Saved to data/processed/review_candidates.txt")
```

```python
# remove unresolvable duplicates: same name AND same club — no way to
# tell the two real players apart with the data available, so both are
# dropped from the direct match to avoid assigning stats to the wrong person
exact_duplicate_names = exact_duplicates['name_normalized'].unique()

df_players_brazil_clean = df_players_brazil[
    ~((df_players_brazil['name_normalized'].isin(exact_duplicate_names)) &
      (df_players_brazil['name_normalized'].isin(matched_players['name_normalized'])))
]
df_players_brazil_clean.shape
```

## Final merge

```python
# direct matches: name_normalized as the join key
direct_matches = df_players_brazil_clean[
    df_players_brazil_clean['name_normalized'].isin(df_stats_by_player['name_normalized'])
][['name_x', 'name_normalized']].copy()
direct_matches['stats_key'] = direct_matches['name_normalized']

# fuzzy matches (club filter + manual): map name_x -> the stats player's name_normalized
fuzzy_club_matches = df_accepted_matches_reviewed[['name_x', 'candidate']].copy()
fuzzy_club_matches['stats_key'] = fuzzy_club_matches['candidate'].apply(normalize_name)

fuzzy_manual_matches = df_manual_matches[['name_x', 'candidate']].copy()
fuzzy_manual_matches['stats_key'] = fuzzy_manual_matches['candidate'].apply(normalize_name)

all_matches = pd.concat([
    direct_matches[['name_x', 'stats_key']],
    fuzzy_club_matches[['name_x', 'stats_key']],
    fuzzy_manual_matches[['name_x', 'stats_key']],
])
all_matches.shape
```

## Conclusion

- Merged df_players_brazil (Transfermarkt, 1021 players after removing
  unresolvable same-name-same-club duplicates) with performance statistics
  from a secondary source (Eduardo Palmieri, Série A 2024), matched by name
  since no shared ID exists between the two sources.
- Matching combined three techniques: direct normalized-name match (401),
  fuzzy matching with club as a blocking filter and score >= 81 (25,
  manually reviewed), and manual fuzzy matches without club filter to
  catch players who changed clubs between 2024 and now (11, manually
  reviewed).
- Resolved duplicate/ambiguous matches: homonyms with identical name AND
  club were dropped (unresolvable), homonyms matching the same stats
  candidate were dropped from the match set, and one specific case
  (Weverton, Cruzeiro vs Palmeiras) was manually resolved after confirming
  the correct club with the user.
- - Final result: 1021 players in the 2024 Série A roster, with ~326 having
  matched performance statistics (goals, assists, xG, xAG, progressive
  passes/carries, defensive actions). The rest are unmatched — some because
  they didn't play in 2024 (reserves, youth, or players who joined later),
  and some due to unresolved name ambiguity (homonyms across different
  clubs, conservatively excluded rather than risking a wrong assignment).
- Known limitation: name-based matching, even with careful validation, may
  contain a small number of undetected errors (swapped homonyms, missed
  transfers). This should be treated as reasonably reliable but not
  perfectly clean data.
- Result exported to data/processed/players_with_stats_2024.csv, ready for
  the Fase 1 cost-benefit scoring and clustering steps.
